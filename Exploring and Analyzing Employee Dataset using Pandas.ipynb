## **Exploring and Analyzing Employee Dataset with Pandas**

#### **Objective:**

##### This project higlights Exploratory Data Analysis (EDA) using Python Pandas for basic data manipulation and apply various data transformation techniques, including handling missing values, filtering, and performing basic data analysis.

##### **Dataset Overview:**

##### The dataset contains the following columns:

##### employee_id: Unique identifier for each employee.

##### name: Name of the employee.

##### age: Age of the employee.

##### department: The department where the employee works (e.g., HR, Engineering, Marketing).

##### salary: Annual salary of the employee.

##### work_experience: Number of years of work experience the employee has.



#### **1. Load the Data**
#### Objective: Load the dataset into a Pandas DataFrame.

```python
import pandas as pd

# Create a basic dataset
data = {
    'employee_id': range(1, 21),
    'name': ['John', 'Jane', 'Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ivy', 'Jack', 'Karen', 'Leo', 'Mia', 'Nina', 'Olivia', 'Paul', 'Quinn', 'Rita'],
    'age': [22, 25, 28, 35, 40, 42, 50, 60, 65, 30, 28, 24, 37, 41, 48, 33, 29, 32, 45, 38],
    'department': ['HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales', 'HR', 'Engineering', 'Marketing', 'Finance', 'Sales'],
    'salary': [50000, 60000, 55000, 75000, 80000, 85000, 100000, 110000, 120000, 65000, 67000, 72000, 85000, 90000, 95000, 48000, 67000, 70000, 80000, 88000],
    'work_experience': [1, 3, 5, 7, 10, 12, 15, 18, 20, 3, 1, 6, 4, 8, 5, 7, 3, 2, 6, 5],
}

# Converting the dataset to a DataFrame
df = pd.DataFrame(data)

# Displaying the first 5 rows of data in the DataFrame 
display(df.head(5))
```

#### **Data Exploration**
#### Objective: Explore the dataset to get an understanding of its structure.

```python
## Check for missing values in the dataset
missing_values = df.isnull()

# Rows with null values
rows_with_nulls = df[missing_values.any(axis=1)]

# Display rows that has null values
display(rows_with_nulls) # result has no null values in the dataset
```

```python
display(missing_values)
```

```python
# Generating the statistic summary of age, salary and work_experience columns
df[['age', 'salary', 'work_experience']].describe()
```

#### **3. Data Transformation**
#### Objective: Perform basic data transformations such as creating new columns and renaming existing ones.


```python
# Creating New Column 'years_to_retirement'
df['years_to_retirement'] = 65 - df['age']
```

```python
# checking the new added column is added and calculated correctly
display(df)
```

```python
# Renaming salary column to annual salary
df.rename(columns ={'salary' :'annual_salary' }, inplace = True)
display(df)
```

```python
# Sort the DataFrame based on work_experience in descending order
sorted_df = df.sort_values (by='work_experience', ascending=False)
display(sorted_df)
```

#### **4. Handling Missing Data**
#### Objective: Handle missing data (if any).

```python
# Randomly pick 3 rows for missing values in annual_salary column
missing_rows = df.sample(n=2, random_state=5).index # If wanted to select the last 3 rows in df then use df.iloc[-3:]

# Replace their salary to None
df.loc[missing_rows, 'annual_salary'] = None

display(df.loc[missing_rows]) # display to review that code applied correctly
```

```python
display(df)
```

```python
# Filling annual_salary null values with mean salary based on the employee department, since salary level differs by department and more realistic. Also to keep the size of the dataset.
# Add a new column for the annual_salary_null filled. this is added for reference to review changes made.

df['annual_salary'] = df.groupby('department') ['annual_salary'].transform(lambda x: x.fillna(x.mean()))
df['annual_salary'] = df['annual_salary'].round(2)

display(df)
```

```python
##  Dropping Null Values
rows_before = len(df) #Checking the no. of rows
print(f"number of rows before dropping nulls: {rows_before}")

df_dropped=df.dropna()

rows_after = len(df_dropped) #checking the no. of rows after dropping null values
print(f"number of rows after dropping nulls: {rows_after}")

display(df_dropped)
```

#### **5. Data Filtering and Selection**
#### Objective: To apply filtering and selecting data based on conditions.

```python
# Filter employees who are older than 30
filtered_df = df[df['age']>30]

# Select columns name, age and annual_salary
selected_df = filtered_df [['name','age','annual_salary']]
display(selected_df)
```

#### **6. Data Aggregation**
#### Objective: Perform group-based aggregation.

```python
## Average Salary per Department

department_avg_salary = df.groupby('department')['annual_salary'].mean().reset_index()
department_avg_salary = department_avg_salary.round(2).rename(columns={'annual_salary': 'average_salary'})

display(department_avg_salary)
```
